# XGBoost

```python
xgb_model=XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)
```

```python
# Removing any [, ], <, >, =

for col in X_train.columns:
    if "[" in col or "]"in col or "<" in col or  "=" in col:
        print(col)
        X_train.columns=(X_train.columns
            .str.replace("[", "", regex=False)
            .str.replace("]", "", regex=False)
            .str.replace("<", "less than",  regex=False)
            .str.replace(">", "greater than", regex=False)
            .str.replace("=", "equal to", regex=False)
        )
        X_test.columns=(X_test.columns
            .str.replace("[", "", regex=False)
            .str.replace("]", "", regex=False)
            .str.replace("<", "less than", regex=False)
            .str.replace(">", "greater than", regex=False)
            .str.replace("=", "equal to", regex=False)
        )
```

```python
xgb_model.fit(X_train, y_train)
```

```python
y_pred_xgb=xgb_model.predict(X_test)
```

### Feature Importance

```python
feature_importance_xgb=pd.DataFrame({"Features":X_train.columns, "Feature Importance":xgb_model.feature_importances_}).sort_values(by="Feature Importance", ascending=False)
feature_importance.head(10)
```

```python
comparison_xgb=pd.DataFrame({"Actual":y_test, "Predicted":y_pred_xgb})
comparison_xgb.to_csv(r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\Dataset\Processed\Comparison_xgb.csv")
```

```python
joblib.dump(xgb_model, r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\models\XGB_model.joblib")
print("Model Saved successfully !!!")
```
