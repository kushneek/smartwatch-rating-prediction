# Smartwatch Rating Prediction: Model Training

### Objective

- The objective of this notebook is to train multiple regression algorithms on the processed smartwatch dataset and compare their performance.

### Algorithms

- Linear Regression
- Decision Tree
- Ridge
- Lasso
- KNN
- Random Forest
- Gradient Boosting
- XGBoost

```python
import pandas as pd
import numpy as np
import joblib

from pathlib import Path
from xgboost import XGBRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

EXPERIMENTS_DIR = Path.cwd().parent / "experiments" / "baseline" / "models"
```

### Load Processed Dataset

```python
SPLITS_DIR = Path.cwd().parent / "Dataset" / "split"
COMPARISON_DIR = Path.cwd().parent / "experiments" / "baseline" / "comparison"

X_train = pd.read_csv(SPLITS_DIR / "X_train.csv")
X_test = pd.read_csv(SPLITS_DIR / "X_test.csv")
y_train = pd.read_csv(SPLITS_DIR / "y_train.csv").squeeze()
y_test = pd.read_csv(SPLITS_DIR / "y_test.csv").squeeze()

#Scaled Data
X_train_scaled = pd.read_csv(SPLITS_DIR / "X_train_scaled.csv")
X_test_scaled = pd.read_csv(SPLITS_DIR / "X_test_scaled.csv")
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

# LinearRegression

```python
model=LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
coef_df=pd.DataFrame({"feature":X_train.columns, "Coefficient": model.coef_})
coef_df= coef_df.sort_values(by='Coefficient', ascending=False)
coef_df.head(10)
```

```python
y_test
```

```python
comparison_LR=pd.DataFrame({"Actual Ratings":y_test, "Predicted Ratings": y_pred})
```

```python
comparison_LR.head(10)
```

```python
joblib.dump(model,  EXPERIMENTS_DIR / "Linear_Regression.joblib")

print('Linear Regression Model Saved Successfully !!!')
```

```python
comparison_LR.to_csv(COMPARISON_DIR / "Comparison_Linear_Regression.csv", index=False)
```

# XGBoost

```python
xgb_model=XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    random_state=42
)
```

```python
# Removing any [, ], <, >, =

for col in X_train.columns:
    if "[" in col or "]"in col or "<" in col or  "=" in col:
        print(col)
        X_train.columns=(X_train.columns
            .str.replace("[", "", regex=False)
            .str.replace("]", "", regex=False)
            .str.replace("<", "less than",  regex=False)
            .str.replace(">", "greater than", regex=False)
            .str.replace("=", "equal to", regex=False)
        )
        X_test.columns=(X_test.columns
            .str.replace("[", "", regex=False)
            .str.replace("]", "", regex=False)
            .str.replace("<", "less than", regex=False)
            .str.replace(">", "greater than", regex=False)
            .str.replace("=", "equal to", regex=False)
        )
```

```python
xgb_model.fit(X_train, y_train)
```

```python
y_pred_xgb=xgb_model.predict(X_test)
```

### Feature Importance

```python
feature_importance_xgb=pd.DataFrame({"Features":X_train.columns, "Feature Importance":xgb_model.feature_importances_}).sort_values(by="Feature Importance", ascending=False)
feature_importance.head(10)
```

```python
comparison_xgb=pd.DataFrame({"Actual":y_test, "Predicted":y_pred_xgb})
comparison_xgb.to_csv(COMPARISON_DIR / "Comparison_xgb.csv", index=False)
```

```python
joblib.dump(xgb_model,  EXPERIMENTS_DIR /"XGB_model.joblib")
print("Model Saved successfully !!!")
```
