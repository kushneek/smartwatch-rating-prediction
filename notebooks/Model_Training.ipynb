```python
import pandas as pd
import numpy as np
import joblib 

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
```

```python
X_train=pd.read_csv(r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\Dataset\Processed\X_train.csv")
X_test=pd.read_csv(r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\Dataset\Processed\X_test.csv")
y_train=pd.read_csv(r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\Dataset\Processed\y_train.csv").squeeze()
y_test=pd.read_csv(r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\Dataset\Processed\y_test.csv").squeeze()
```

```python
scaler=StandardScaler()

X_train_scaled=pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
X_test_scaled=pd.DataFrame(scaler.transform(X_test), columns=X_test.columns)
```

```python
print(X_train.shape)
print(X_test.shape)

print(y_train.shape)
print(y_test.shape)
```

# LinearRegression

```python
model=LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
coef_df=pd.DataFrame({"feature":X_train.columns, "Coefficient": model.coef_})
coef_df= coef_df.sort_values(by='Coefficient', ascending=False)
coef_df.head(10)
```

```python
y_test
```

```python
comparison_LR=pd.DataFrame({"Actual Ratings":y_test, "Predicted Ratings": y_pred})
```

```python
comparison_LR.head(10)
```

```python
joblib.dump(model, r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\models\Linear_Regression.joblib")

print('Linear Regression Model Saved Successfully !!!')
```

```python
comparison_LR.to_csv(r"C:\Users\ASUS\OneDrive\Documents\ML Project\Smartwatches project\Dataset\Processed\Comparison_Linear_Regression.csv", index=False)
```
